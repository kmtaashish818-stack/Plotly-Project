```python
import numpy as np
import pandas as pd
```

```python
latlong=pd.read_csv('indian geography\district wise centroids.csv')
```

```python
latlong.head()
```

```python
latlong.shape
```

```python
census=pd.read_csv('india census 2011\india-districts-census-2011.csv')
```

```python
census.head()
```

```python
census.shape
```

```python
cols=['District code','District name','Population','Male','Female','Literate','Households_with_Internet','Housholds_with_Electric_Lighting']
```

```python
census=census[cols]
```

```python
final_df=latlong.merge(census,left_on='District',right_on='District name').drop(columns=['District name'])
final_df
```

```python
final_df['sex_ratio']=round((final_df['Female']/final_df['Male'])*100)
```

```python
final_df['literacy_rate']=round((final_df['Literate']/final_df['Population'])*100)
```

```python
final_df.head()
```

```python
final_df=final_df.drop(columns=['Male','Female','Literate'])
```

```python
final_df.to_csv('india.csv',index=False)
```
